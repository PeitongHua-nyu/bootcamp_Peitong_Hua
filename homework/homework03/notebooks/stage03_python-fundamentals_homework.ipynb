## 1. NumPy Operations

```python
arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("Add 10:", arr + 10)
print("Multiply by 2:", arr * 2)
print("Square:", arr ** 2)
```

## 2. Compare Loop vs Vectorization

```python
arr = np.arange(10000)

def loop_square(arr):
    result = []
    for x in arr:
        result.append(x ** 2)
    return result

%timeit loop_square(arr)
%timeit arr ** 2
```
